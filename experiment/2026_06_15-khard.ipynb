```python
with open("../../monochrome/module/person/contacts.vcf",
	"r") as contacts_file:
	for line in contacts_file:
		line = line.strip()
		if line == "BEGIN:VCARD":
			uid = str(uuid.uuid4()).replace("-", "")
			individual_file = open(
				f"../../monochrome/module/vcard/{uid}.vcf", "w")
			individual_file.write(f"{line}\n")
			individual_file.write(f"UID:{uid}\n")
		elif line == "END:VCARD":
			individual_file.write(f"{line}\n")
			individual_file.close()
		elif not individual_file.closed:
			individual_file.write(f"{line}\n")
```
