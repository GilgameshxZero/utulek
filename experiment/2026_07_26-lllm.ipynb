```python
from transformers import pipeline
```

```python
pipe = pipeline(task="text-generation",
	model="Qwen/Qwen3.5-4B")
pipe("Which is greater, 3.9 or 3.11?")
```

```python
pipe("How many \"r\"'s are there in \"strawberry\"?")
```

```python
pipe(
	"Write a C++ program to compute the N-th prime number.")
```

```python
res = pipe(
	"Write a C++ program to compute the N-th prime number.",
	max_new_tokens=65536)
print(res[0]["generated_text"])
```

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained(
	"Qwen/Qwen3.5-4B")
chat = [
	{
	"role": "user",
	"content": "Hello, how are you?"
	},
	{
	"role": "assistant",
	"content": "I'm doing great. How can I help you today?"
	},
	{
	"role":
	"user",
	"content":
	"I'd like to show off how chat templating works!"
	},
]

tokenizer.apply_chat_template(chat, tokenize=False)
```

```python
chat = [
	{
	"role": "system",
	"content": "Assist the user in their task."
	},
	{
	"role": "user",
	"content": "Analyze the 2026 horror movie \"Obsession\"."
	},
]
res = pipe(text_inputs=tokenizer.apply_chat_template(chat,
	tokenize=False),
	max_new_tokens=2048)
pp(res[0]["generated_text"])
```
